# M8_8.20 · Entornos, Conda, contenedores y versiones

## Entorno virtual

Un entorno virtual aísla el intérprete y las dependencias de un proyecto. Evita que una actualización necesaria para un proyecto rompa otro. **Conda** gestiona entornos y paquetes; `pip` instala paquetes en el entorno activo. Un archivo `environment.yml` documenta el entorno.

Comandos de referencia, para ejecutar en terminal, no dentro de esta celda:

```bash
conda create -n modulo8 python=3.12
conda activate modulo8
conda install pandas numpy matplotlib openpyxl jupyter
conda env export --from-history > environment.yml
conda env create -f environment.yml
```

Antes de instalar, comprobar siempre qué Python está activo:

```bash
which python      # macOS/Linux
where python      # Windows
python --version
```

````python
import sys, platform
print("Python:",sys.version.split()[0])
print("Ejecutable:",sys.executable)
print("Sistema:",platform.system())
````

## Contenedores y Docker

Un contenedor empaqueta una aplicación y su entorno de ejecución. Docker es una plataforma habitual para construir y ejecutar contenedores. Un entorno Conda aísla paquetes de Python; un contenedor puede aislar además componentes del sistema operativo. En este módulo se requiere reconocer el propósito, no construir imágenes Docker.

## Git y GitHub

**Git** es un sistema de control de versiones: registra instantáneas llamadas *commits*, permite comparar cambios y recuperar estados anteriores. **GitHub** aloja repositorios Git y facilita compartir y colaborar.

Conceptos mínimos:

- repositorio: carpeta de proyecto con historial;
- commit: instantánea con un mensaje;
- branch: línea de trabajo;
- remote: repositorio remoto, normalmente `origin`;
- clone: crear una copia completa del repositorio;
- pull: incorporar cambios remotos;
- push: publicar commits locales.

Flujo local de referencia:

```bash
git clone https://github.com/USUARIO/m8-herramientas-data-science.git
cd m8-herramientas-data-science
git status
git add M8_8.19_21/notebook.ipynb
git commit -m "Completa ejercicios de Python"
git push
```

## Estructura del repositorio

```text
m8-herramientas-data-science/
├── data/input/       datos originales
├── data/database/    SQLite docente
├── data/output/      resultados regenerables
├── M8_8.xx/          un directorio por tarde
├── docs/             coordinación y referencias
├── environment.yml
└── README.md
```

`data/input` no se sobrescribe. Los resultados se escriben en `data/output`. Los notebooks viven en la carpeta de su tarde.

````python
from pathlib import Path
import os, sys, subprocess

REPO_NAME = "m8-herramientas-data-science"
GITHUB_USER = "REPLACE_WITH_YOUR_GITHUB_USERNAME"  # El profesorado lo cambia una vez antes de publicar

def localizar_repo():
    actual = Path.cwd().resolve()
    for candidato in [actual, *actual.parents]:
        if (candidato / "data" / "input").exists():
            return candidato
    if "google.colab" in sys.modules:
        destino = Path("/content") / REPO_NAME
        if not destino.exists():
            url = f"https://github.com/{GITHUB_USER}/{REPO_NAME}.git"
            if GITHUB_USER.startswith("REPLACE_"):
                raise RuntimeError("El profesorado debe configurar GITHUB_USER antes de publicar el repositorio.")
            subprocess.run(["git", "clone", "--depth", "1", url, str(destino)], check=True)
        return destino
    raise FileNotFoundError("No se encuentra la raiz del repositorio.")

ROOT = localizar_repo()
DATA = ROOT / "data" / "input"
DB = ROOT / "data" / "database" / "hidrogeologia.sqlite"
OUTPUT = ROOT / "data" / "output"
for carpeta in [OUTPUT / "tables", OUTPUT / "figures", OUTPUT / "logs"]:
    carpeta.mkdir(parents=True, exist_ok=True)
print("Repositorio:", ROOT)
````

## Condiciones y bucles

Las condiciones seleccionan acciones; los bucles repiten una operación. Deben responder a una necesidad real y no sustituir operaciones vectorizadas disponibles más adelante.

````python
for valor in [8.7,-1,24.5]:
    if valor < 0:
        print(valor,"valor imposible")
    elif valor > 20:
        print(valor,"nivel profundo")
    else:
        print(valor,"nivel somero o intermedio")
````

## Funciones, scripts y módulos

Una función recibe argumentos y devuelve un resultado. Un módulo es un archivo `.py` importable. El patrón `if __name__ == "__main__":` separa la ejecución principal del código reutilizable. Las clases y `self` se presentan solo como conceptos de reconocimiento.

````python
def convertir_a_cota(cota_terreno_m,profundidad_m):
    """Calcula cota piezometrica; devuelve NaN si la profundidad no es valida."""
    if profundidad_m is None or profundidad_m < 0:
        return np.nan
    return cota_terreno_m-profundidad_m
[convertir_a_cota(112.4,x) for x in [8.7,-1,None]]
````
