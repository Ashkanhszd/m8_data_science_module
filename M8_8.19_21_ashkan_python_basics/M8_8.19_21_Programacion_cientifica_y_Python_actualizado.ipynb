# M8_8.19 · Programación científica

## 1. Qué significa programar

Programar es expresar una secuencia de instrucciones que un ordenador puede ejecutar. En programación científica, el código forma parte del razonamiento y debe permitir:

1. recibir datos;
2. comprobarlos;
3. transformarlos;
4. producir resultados;
5. revisar los resultados;
6. conservar las decisiones y el procedimiento.

Un programa que termina sin error no es necesariamente científicamente correcto.

## 2. Entrada, proceso, salida y comprobación

Un flujo sencillo puede expresarse como:

```text
entrada → validación → cálculo → salida → comprobación
```

Ejemplo hidrogeológico: calcular la cota piezométrica a partir de la cota del terreno y la profundidad del nivel.

````python
# Entrada
cota_terreno_m = 112.4
profundidad_nivel_m = 8.7

# Validación sencilla
if profundidad_nivel_m < 0:
    raise ValueError("La profundidad del nivel no puede ser negativa.")

# Cálculo
cota_piezometrica_m = cota_terreno_m - profundidad_nivel_m

# Salida
print("Cota piezométrica:", cota_piezometrica_m, "m")

# Comprobación
print(
    "¿La cota piezométrica es inferior a la cota del terreno?",
    cota_piezometrica_m < cota_terreno_m
)
````

## 3. Algoritmo y pseudocódigo

Un **algoritmo** es una secuencia finita y ordenada de pasos para resolver un problema. El **pseudocódigo** describe esos pasos con lenguaje estructurado, sin depender de la sintaxis de Python.

```text
RECIBIR cota del terreno
RECIBIR profundidad del nivel
SI la profundidad es negativa
    INFORMAR de un error
SI NO
    CALCULAR cota = cota del terreno - profundidad
    MOSTRAR el resultado
FIN
```

Escribir primero el pseudocódigo ayuda a separar el razonamiento científico de la sintaxis.

## 4. Notebook, script, aplicación y API

### Notebook
Documento interactivo formado por celdas de texto, código y resultados. Es apropiado para aprender, explorar datos y documentar un análisis. El orden de ejecución importa.

### Script
Archivo de texto, normalmente con extensión `.py`, que contiene instrucciones Python. Es apropiado para procesos repetibles y automatización.

### Aplicación
Programa diseñado para que una persona realice una tarea mediante una interfaz o un flujo definido.

### API
Una **interfaz de programación de aplicaciones** define cómo un programa solicita datos o funciones a otro componente. Una API de modelización puede permitir crear, ejecutar o leer un modelo desde Python.

### Frontend y backend
El **frontend** es la parte con la que interactúa el usuario. El **backend** realiza operaciones, gestiona datos o responde a solicitudes. Estos conceptos se presentan a nivel de reconocimiento.

## 5. Jupyter y Google Colab

Jupyter ejecuta código mediante un **kernel**. El kernel conserva en memoria las variables creadas durante la sesión.

Google Colab ofrece un entorno Jupyter remoto. Una sesión de Colab es temporal:

- reiniciar el entorno borra las variables;
- los archivos guardados en el almacenamiento temporal pueden desaparecer;
- conviene guardar una copia del notebook en Drive o descargarla;
- ejecutar celdas fuera de orden puede producir resultados difíciles de reproducir.

Una comprobación útil es reiniciar el entorno y ejecutar todas las celdas desde el principio.

## 6. Errores habituales

- **`SyntaxError`**: la sintaxis no es válida.
- **`NameError`**: se utiliza un nombre que no existe en memoria.
- **`TypeError`**: una operación no es compatible con el tipo de dato.
- **`ValueError`**: el tipo es aceptable, pero el valor no.
- **`FileNotFoundError`**: no se encuentra un archivo o ruta.

Los errores deben corregirse, no ocultarse. El mensaje es parte del proceso de diagnóstico.

````python
# Ejemplos controlados de errores y lectura del mensaje.
ejemplos = [
    ("NameError", "Usar una variable que no existe"),
    ("TypeError", "Sumar texto y número"),
    ("ValueError", "Convertir texto no numérico a entero")
]

for nombre, explicacion in ejemplos:
    print(nombre, "→", explicacion)
````

## 7. Intérprete, terminal, IDE y notebook

- **Intérprete de Python:** programa que ejecuta instrucciones Python.
- **Terminal:** interfaz textual para ejecutar comandos del sistema.
- **IDE:** aplicación para escribir, ejecutar y depurar código.
- **Notebook:** documento interactivo conectado a un kernel.

Un notebook puede estar abierto en un navegador, pero el código se ejecuta mediante un intérprete situado en un entorno concreto.

````python
import sys
import platform

print("Versión de Python:", sys.version.split()[0])
print("Ejecutable de Python:", sys.executable)
print("Sistema operativo:", platform.system())
````

## 8. Paquetes, dependencias y versiones

Un **paquete** reúne código reutilizable. Una **dependencia** es un paquete que necesita el proyecto. La versión importa porque las funciones y comportamientos pueden cambiar.

```python
import numpy as np
import pandas as pd
```

El alias `np` o `pd` es una convención. No instala el paquete: solo lo importa en la sesión activa.

````python
import numpy as np

print("Versión de NumPy:", np.__version__)
````

## 9. Entornos virtuales y Conda

Un **entorno virtual** aísla el intérprete y los paquetes de un proyecto. Evita que proyectos distintos dependan de una única instalación global.

**Conda** puede crear entornos e instalar paquetes. `pip` instala paquetes Python en el entorno activo.

Comandos de referencia para una terminal:

```bash
conda create --name modulo8 python=3.12
conda activate modulo8
conda install numpy pandas matplotlib openpyxl jupyter
conda deactivate
```

Un archivo `environment.yml` documenta el entorno:

```yaml
name: modulo8
channels:
  - conda-forge
dependencies:
  - python=3.12
  - numpy
  - pandas
  - matplotlib
  - openpyxl
  - jupyter
```

Para reconstruirlo:

```bash
conda env create --file environment.yml
```

Los comandos de terminal no se escriben como Python ordinario.

## 10. Entorno virtual y contenedor

Un entorno Conda aísla principalmente Python y sus dependencias. Un **contenedor** puede empaquetar además bibliotecas y componentes del sistema operativo.

**Docker** es una plataforma habitual para construir y ejecutar contenedores.

```text
Conda: proyecto → Python + paquetes
Contenedor: proyecto → aplicación + dependencias + entorno del sistema
```

En este módulo, los contenedores se presentan para reconocer su finalidad. No es necesario crear una imagen Docker.

# M8_8.21 · Fundamentos de Python

## 15. Variables y asignación

Una variable asocia un nombre con un valor. `=` asigna; no significa igualdad matemática.

Los nombres deben ser descriptivos, no comenzar con número y no contener espacios. Python distingue mayúsculas y minúsculas.

````python
id_pozo = "P01"
cota_terreno_m = 112.4
profundidad_nivel_m = 8.7
medicion_valida = True

print(id_pozo)
print(cota_terreno_m)
print(profundidad_nivel_m)
print(medicion_valida)
````

## 16. Tipos básicos

- **`int`**: entero.
- **`float`**: número decimal.
- **`str`**: texto.
- **`bool`**: `True` o `False`.
- **`None`**: ausencia de un objeto o resultado.
- **`NaN`**: valor numérico ausente, frecuente en NumPy y pandas.

`None`, `NaN`, cero y cadena vacía no son equivalentes.

````python
import numpy as np

valores = [42, 8.7, "P01", True, None, np.nan]

for valor in valores:
    print(repr(valor), "→", type(valor))
````

## 17. Conversión de tipos

Las funciones `int()`, `float()`, `str()` y `bool()` convierten tipos cuando la conversión es válida.

````python
profundidad_texto = "8.7"
profundidad_numero = float(profundidad_texto)

print("Antes:", profundidad_texto, type(profundidad_texto))
print("Después:", profundidad_numero, type(profundidad_numero))
````

## 18. Operadores

### Aritméticos
`+`, `-`, `*`, `/`, `**`, `%`

### Comparación
`==`, `!=`, `<`, `<=`, `>`, `>=`

### Lógicos
`and`, `or`, `not`

Una comparación produce un booleano. `=` asigna; `==` compara.

````python
cota_piezometrica_m = cota_terreno_m - profundidad_nivel_m
es_nivel_profundo = profundidad_nivel_m > 20
esta_en_rango = profundidad_nivel_m >= 0 and profundidad_nivel_m <= 100

print("Cota piezométrica:", cota_piezometrica_m)
print("¿Nivel profundo?", es_nivel_profundo)
print("¿Valor dentro del rango?", esta_en_rango)
````

## 19. Cadenas de texto

Una cadena almacena texto. Las *f-strings* integran valores en mensajes legibles.

````python
mensaje = (
    f"El pozo {id_pozo} tiene una profundidad del nivel de "
    f"{profundidad_nivel_m:.2f} m."
)

print(mensaje)
````

## 20. Listas y tuplas

Una **lista** es una colección ordenada y modificable. Una **tupla** es ordenada y normalmente se utiliza para valores que no deben cambiar.

Los índices comienzan en cero.

````python
niveles = [8.7, 8.4, 8.9]
coordenadas = (421100, 4582100)

print("Primer nivel:", niveles[0])
print("Último nivel:", niveles[-1])

niveles.append(9.1)
print("Lista modificada:", niveles)
print("Coordenadas:", coordenadas)
````

## 21. Diccionarios

Un diccionario almacena pares clave-valor. Es útil para representar una entidad pequeña o una configuración. Las claves deben ser únicas.

````python
pozo = {
    "id_pozo": "P01",
    "acuifero": "Aluvial",
    "cota_terreno_m": 112.4
}

print("Identificador:", pozo["id_pozo"])
print("Acuífero:", pozo["acuifero"])

pozo["activo"] = True
print(pozo)
````

## 22. Condicionales

`if`, `elif` y `else` ejecutan bloques según condiciones. La indentación define el bloque y forma parte de la sintaxis.

````python
profundidad = 24.5

if profundidad < 0:
    clasificacion = "valor imposible"
elif profundidad < 10:
    clasificacion = "nivel somero"
elif profundidad < 20:
    clasificacion = "nivel intermedio"
else:
    clasificacion = "nivel profundo"

print(clasificacion)
````

## 23. Bucles `for`

Un bucle `for` recorre elementos. Es adecuado cuando la misma acción debe aplicarse a varios objetos. En M8_8.22 se verá que NumPy puede evitar muchos bucles numéricos mediante vectorización.

````python
niveles = [8.7, 8.4, 8.9]

for posicion, nivel in enumerate(niveles, start=1):
    print("Observación", posicion, "→", nivel, "m")
````

## 24. Funciones

Una función agrupa una tarea reutilizable.

- `def` inicia la definición.
- Los **parámetros** son nombres declarados en la función.
- Los **argumentos** son valores utilizados al llamarla.
- `return` devuelve un resultado.
- Una **docstring** explica finalidad, entradas y salida.

Una función debería hacer una tarea clara y no depender innecesariamente de variables externas.

````python
def calcular_cota_piezometrica(cota_terreno_m, profundidad_nivel_m):
    """Calcula la cota piezométrica a partir de medidas en metros."""

    if profundidad_nivel_m is None:
        return np.nan

    if profundidad_nivel_m < 0:
        raise ValueError("La profundidad no puede ser negativa.")

    cota_piezometrica_m = cota_terreno_m - profundidad_nivel_m
    return cota_piezometrica_m


resultado = calcular_cota_piezometrica(
    cota_terreno_m=112.4,
    profundidad_nivel_m=8.7
)

print("Resultado:", resultado, "m")
````

## 25. Alcance de variables

Una variable creada dentro de una función es local. Utilizar parámetros y valores devueltos hace el código más fácil de probar y reutilizar.

````python
def duplicar(valor):
    resultado_local = valor * 2
    return resultado_local


resultado_externo = duplicar(4)
print(resultado_externo)
````

## 26. Scripts, módulos e importaciones

Un archivo `.py` puede ser:

- **script**, si se ejecuta como proceso;
- **módulo**, si se importa para reutilizar funciones.

```python
from pathlib import Path
from mi_modulo import calcular_cota_piezometrica
```

El patrón siguiente separa funciones y ejecución principal:

```python
def main():
    resultado = calcular_cota_piezometrica(112.4, 8.7)
    print(resultado)

if __name__ == "__main__":
    main()
```

No es necesario dominar este patrón todavía. Se retomará al organizar flujos reproducibles.
